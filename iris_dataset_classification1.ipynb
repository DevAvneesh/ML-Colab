```python
import numpy as np
from sklearn import datasets
```

```python
from sklearn.datasets import load_iris
df = load_iris()
df
```

```python
import pandas as pd
import numpy as np
dataframe = pd.DataFrame(np.c_[df['data'], df['target']], columns = df['feature_names'] + ['target'])
dataframe.head()
```

```python
dataframe.describe()
```

```python
import matplotlib.pyplot as plt
dataframe.plot(kind = 'scatter', x = 'sepal length (cm)', y = 'sepal width (cm)')
plt.show()
```
